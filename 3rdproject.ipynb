```python
x=df.drop(['price'],axis=1)
y=df['price']
```

```python
x
```

```python
df.columns
```

```python
x_one_encode = pd.get_dummies(x,columns=['model','transmission','fuelType'],drop_first=True)
```

```python
x_one_encode=x_one_encode.astype(int)
```

```python
x_one_encode
```

```python
from sklearn.preprocessing import LabelEncoder
encoder=LabelEncoder()
```

```python
columns = ['model', 'transmission', 'fuelType']
```

```python
xlabel=x
for i in columns:
    xlabel[i]=encoder.fit_transform(xlabel[i])
```

```python
xlabel
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
numeric_columns = ['year', 'mileage', 'tax', 'mpg', 'engineSize']
scaler = StandardScaler()
x_one_encode[numeric_columns] = scaler.fit_transform(x_one_encode[numeric_columns])
```

```python
x_one_encode
```

```python
import pandas as pd
df_fixed=pd.DataFrame(xlabel,columns=['model','year','transmission','mileage','fuelType','tax','mpg','enginesize'])
```

```python
df_fixed.columns
```

```python
xlabel[['model','year','transmission','mileage','fuelType','tax','mpg','engineSize']]=scaler.fit_transform(xlabel[['model','year','transmission','mileage','fuelType','tax','mpg','engineSize']])
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
x_train, x_test, y_train, y_test = train_test_split(xlabel, y, test_size=0.2, random_state=42)
```

```python
model = LinearRegression()
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
y_pred
```

```python
y_test
```

```python
r2_score(y_test, y_pred)
```

```python
n=x_test.shape[0]
p=x_test.shape[1]
adjusted_r2 = 1 - (1 - r2_score(y_test, y_pred)) * (n - 1) / (n - p - 1)
print("Adjusted R-squared:", adjusted_r2)
```
